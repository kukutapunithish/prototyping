```python
from pydantic import BaseModel, ValidationError

class User(BaseModel):
    username: str
    age: int
    is_active: bool = True  # Default value
```

```python
# This works perfectly; age is converted from string '25' to integer 25
user_data = {"username": "jdoe_99", "age": "25"}
user = User(**user_data)
```

```python
print(user.model_dump()) 
print(user.model_dump_json()) 
```

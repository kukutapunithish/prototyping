## Implementing normal linked list

```python
class Node:
    def __init__(self, data):
        self.data = data
        self.next = None

class LinkedList:
    def __init__(self):
        self.head = None

    def push(self,new_data):
        new_node = Node(new_data)
        new_node.next = self.head
        self.head = new_node

    def insertAfter(self,prev_node,new_data):
        if prev_node is None:
            print('The given node is not present in linkedlist')
            return
        new_node = Node(new_data)
        new_node.next = prev_node.next
        prev_node.next = new_node

    def append(self,new_data):
        new_node = Node(new_data)
        if self.head is None:
            self.head = new_node
            return

        last = self.head
        while(last.next):
            last = last.next
        last.next = new_node

    def printList(self):
        temp = self.head
        while (temp):
            print(temp.data,end=' => ')
            temp = temp.next
```

```python
# Start with the empty list
llist = LinkedList()
```

```python
class Node:
    def __init__(self, data, level):
        self.data = data

        # Multiple forward pointers.
        # next[0] -> level 0
        # next[1] -> level 1
        # next[2] -> level 2
        self.next = [None] * level

class SkipList:
    def __init__(self, number_of_levels=3):
        self.number_of_levels = number_of_levels

        # Head is a dummy node.
        # It does not contain actual data.
        self.head = Node(None, number_of_levels)

    def insert(self, data, node_level):
        """
        Insert a node with a fixed height.

        node_level = number of levels this node participates in.
        Example:
            node_level=1 -> level 0 only
            node_level=2 -> levels 0 and 1
            node_level=3 -> levels 0, 1 and 2
        """

        # Don't allow node height to exceed total levels.
        node_level = min(node_level, self.number_of_levels)

        new_node = Node(data, node_level)

        # We search from the highest level down.
        current = self.head

        # Keep track of the previous node at every level.
        update = [None] * self.number_of_levels

        for level in range(self.number_of_levels - 1, -1, -1):

            while (
                current.next[level] is not None
                and current.next[level].data < data
            ):
                current = current.next[level]

            update[level] = current

        # Connect the new node into every level it belongs to.
        for level in range(node_level):
            new_node.next[level] = update[level].next[level]
            update[level].next[level] = new_node


    def search(self, target):
        current = self.head

        # Start from the highest level.
        for level in range(self.number_of_levels - 1, -1, -1):

            # Move right while we can.
            while (
                current.next[level] is not None
                and current.next[level].data <= target
            ):
                current = current.next[level]

        # After all levels, current should be the target
        # or the greatest value smaller than target.
        if current.data == target:
            return current

        return None


    def print_levels(self):
        for level in range(self.number_of_levels - 1, -1, -1):

            current = self.head.next[level]

            print(f"Level {level}: ", end="")

            while current:
                print(current.data, end=" -> ")
                current = current.next[level]

            print("None")
```

```python
# %%
skip_list = SkipList(number_of_levels=3)

# data, number of levels
skip_list.insert(20, 3)
skip_list.insert(23, 1)
skip_list.insert(25, 3)
skip_list.insert(26, 1)
skip_list.insert(31, 2)
skip_list.insert(42, 1)

skip_list.print_levels()

print(skip_list.search(31).data)
```
